```python
from pathlib import Path

data_path = Path("data/names.txt")

words = data_path.read_text().splitlines()

print("number of names:", len(words))
print("first 10 names:", words[:10])
print("shortest name:", min(words, key=len))
print("longest name:", max(words, key=len))
print("total characters:", sum(len(w) for w in words))
```

```python
all_words = ""
for word in words:
    all_words += word
unique = sorted(list(set(all_words)))
unique
```

```python
counts = {}
for word in words:
    word = "." + word + "."
    for i in range(len(word) - 1):
        counts[(word[i], word[i+1])] = counts.get((word[i], word[i+1]), 0) +  1
print(counts)
```

```python
stoi = {}
itos = {}
unique = ["."] + unique
for i, ch in enumerate(unique):
    stoi[ch] = i
    itos[i] = ch
print(stoi)
print(itos)
```

```python
import numpy as np

# np.zeros?

N = np.zeros((len(unique), len(unique)), np.int16)
N.shape
```

```python
counts
```

```python
for pair, count in counts.items():
    N[stoi[pair[0]], stoi[pair[1]]] = count
N
```

```python
row_sum = N.sum(axis=1, keepdims=True)
P = N / row_sum
P.sum(axis=1)
```

```python
with torch.no_grad():
    for _ in range(10):
        word = ""
        char = "."
        while True:
            char_one_hot = torch.nn.functional.one_hot(torch.tensor(stoi[char]), num_classes=27).double()
            result = char_one_hot @ W_torch
            logits = torch.softmax(result, dim=0)
            char = torch.multinomial(logits, num_samples=1)
            char = itos[char.item()]
            word = word + char
            if char == ".":
                break
        print(word)
```
